```python

from langchain_community.llms import OpenAI
from dotenv import load_dotenv

load_dotenv()

def generate_cat_name():
    llm = OpenAI(temperature=0.6)
    names = llm("Voce tem um gatinho filho novo e gostaria de da rum nome legal para ele. Me de uma lista de 5 possiveis nomes.")
    return names

if __name__ == "__main__":
    print(generate_cat_name())
```
